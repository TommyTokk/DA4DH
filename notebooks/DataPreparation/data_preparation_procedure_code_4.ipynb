```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
import sys
import os
sys.path.append(os.path.abspath(os.path.join(os.getcwd(), '../..')))

from src.utils import data_understanding as du
from src.utils import utils as ut


procedure_code_df = pd.read_csv('../../data/raw/procedure_code_4.csv')
```

```python
# For each subject_id, hadm_id, count the number of unique procedure codes
procedure_counts = procedure_code_df.groupby(['subject_id', 'hadm_id'])['long_title'].nunique().reset_index()
procedure_counts.rename(columns={'long_title': 'number_unique_procedures'}, inplace=True)

# For each subject_id, hadm_id, get the ICD9 codes
icd9_codes = procedure_code_df.groupby(['subject_id', 'hadm_id'])['icd_code'].apply(lambda x: ','.join(x.unique())).reset_index()

# Merge the counts with the ICD9 codes
merged_df = pd.merge(procedure_counts, icd9_codes, on=['subject_id', 'hadm_id'])

merged_df.to_csv('../../data/processed/procedure_code_summary.csv', index=False)
```
